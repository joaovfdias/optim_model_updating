```python
import pandas as pd
import glob
import os

algo = "PSO"

# Define o caminho da pasta onde estão os arquivos CSV.
# './' significa a pasta atual. Altere se os arquivos estiverem em outro diretório.

caminho_saida = os.path.join(r"D:\Thiago Artur\OneDrive\Documentos\2025.2\Pesquisa\Rodadas\Problema 2\META-OPT\9 PARAM", algo)
caminho_pasta = os.path.join(caminho_saida, "arquivos")
titulo_saida = f"meta_opt_{algo}.csv"

# Busca todos os arquivos que terminam com .csv na pasta especificada
arquivos_csv = glob.glob(os.path.join(caminho_pasta, f"meta_opt_{algo}_*.csv"))

# Lista para armazenar o conteúdo de cada arquivo
lista_dfs = []

for arquivo in arquivos_csv:

    # Lê o arquivo CSV atual
    df = pd.read_csv(arquivo)
    lista_dfs.append(df)

if not lista_dfs:
    print("Nenhum arquivo CSV encontrado para unir.")
else:
    # Junta (concatena) todos os arquivos da lista em um único DataFrame
    df_completo = pd.concat(lista_dfs, ignore_index=True)

    # Converte a coluna 'Timestamp' para o tipo datetime para garantir a ordenação cronológica correta
    df_completo['Timestamp'] = pd.to_datetime(df_completo['Timestamp'])

    # Ordena as linhas baseando-se na coluna 'Timestamp' em ordem crescente
    df_completo = df_completo.sort_values(by='Timestamp')

    # Salva o resultado em um novo arquivo CSV, preservando o cabeçalho e sem adicionar a coluna de índices
    df_completo.to_csv(os.path.join(caminho_saida, titulo_saida), index=False)

    print("Sucesso! Todos os arquivos foram unidos e salvos em 'arquivo_unificado.csv'.")
```
